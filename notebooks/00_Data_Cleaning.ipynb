### Data Cleaning

```python
# import necessary libraries
import pandas as pd
```

```python
# load the dataset
df = pd.read_excel('../dataset/Dataset.xlsx')
df.head()
```

```python
# Physical check of last rows
df.tail()
```

```python
# Check for duplicate rows
print("Duplicate rows:", df.duplicated().sum())
```

```python
# Check for missing values
print("\nMissing values:")
print(df.isna().sum())
```

```python
# Handle missing values in the 'CouponCode' column by filling them with 'No Coupon'
df['CouponCode'] = df['CouponCode'].fillna('No Coupon')
```

```python
# Check the value counts of the 'CouponCode' column after handling missing values
df['CouponCode'].value_counts()
```

```python
# Check for missing values again after handling them(Validating the changes)
print(df.isna().sum())
```

```python
# Check for duplicate Order IDs and Tracking Numbers
print("Duplicate Order IDs:", df['OrderID'].duplicated().sum())
print("Duplicate Tracking Numbers:", df['TrackingNumber'].duplicated().sum())
```

```python
# Check the data types of each column
df.dtypes
```

```python
# Check for negative or zero values in Quantity, UnitPrice, TotalPrice, and ItemsInCart columns
print("Quantity <= 0:", (df['Quantity'] <= 0).sum())
print("UnitPrice <= 0:", (df['UnitPrice'] <= 0).sum())
print("TotalPrice <= 0:", (df['TotalPrice'] <= 0).sum())
print("ItemsInCart <= 0:", (df['ItemsInCart'] <= 0).sum())
```

```python
# Check for incorrect TotalPrice values by comparing it with the calculated total (Quantity * UnitPrice)
calculated_total = df['Quantity'] * df['UnitPrice']

print(
    "Incorrect TotalPrice:",
    (abs(df['TotalPrice'] - calculated_total) > 0.01).sum()
)
```

```python
# Check for ItemsInCart values that are less than Quantity values
print(
    "ItemsInCart < Quantity:",
    (df['ItemsInCart'] < df['Quantity']).sum()
)
```

```python
# Check for unique values in categorical columns
print(df['PaymentMethod'].unique())
print(df['OrderStatus'].unique())
print(df['Product'].unique())
print(df['ReferralSource'].unique())
print(df['CouponCode'].unique())
```

```python
# Check the earliest and latest dates in the 'Date' column
print("Earliest date:", df['Date'].min())
print("Latest date:", df['Date'].max())
```
